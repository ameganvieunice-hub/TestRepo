```python
# fonction si
x=0
def sign(x):
    if x>0:
        print(x, "positif")
    elif x==0:
        print(x, "nul")
    else:
        print(x," négatif")
sign(3)
```

```python
villes=['lomé','aného','vogan','Kara']
print(villes[-1]) #le dernier element
print(villes[0]) # le premier element
print(villes[2:]) #à partir du 3e element jusqu'à la fin
print(villes[:3]) #à partir du début jusqu'au 3e element
print(villes[::2]) #du début à la fin en sautant un élément chaque fois
villes.append('Tsévié') #ajoutez un élément à la fin d'une liste
villes
villes.insert(3,'Kpalimé') #ajoutez un élément n'importe où avec l'indexage
villes
villes_2=['Atakpamé','Sokodé']
villes.extend(villes_2) #ajoutez une liste entière à une autre
villes
len(villes) #compter le nombre d'élément à l'intérieur d'une liste
villes.sort() #Trie par ordre alphabétique
villes
villes.sort(reverse=True) #Trie par ordre décroissant
villes
villes[5]='lomé'
villes.count('lomé') #compte le nombre de fois qu'un élément apparaît dans une liste
#mettre les valeurs de la liste avec leur index
for index, valeur in enumerate(villes):
    print(index,valeur)
région=["maritime","maritime","maritime","maritime","Centrale","maritime","Kara","Plateau",]
len(région)
for c,d in zip(villes,région): #associez 2 listes avec zip
    print(c,"est dans la région",d)
```

```python
#définition des variables
name=input(print("Bienvenue chez NICE'S BAG. Veuillez entrer votre nom"))
produits=['Louis Vuitton','Hermès','Chanel','Dior']
prix=[4500,4000,3500,2500]
#création de dictionnaire (à gauche, ce sont les clés(produits). à droite, ce sont les valeurs(prix))
magasin = {
    "Louis Vuitton": 4500,
    'Hermès': 4000,
    'Chanel': 3500,
    'Dior': 2500
    }
magasin["Gucci"]=3000 #ajout d'une clé dans le dictionnaire
#salutation et présentation des produits et prix
print("Bonjour", name,"! Nous disposons des sacs avec leur prix respectif:")
for k, v in magasin.items():
    print(k,v)
choix=input(print('Quel sac désirez-vous?')) #saisie du prdouit que l'utilisateur désire
#vérification du produit dans la base
if choix in produits:
    print(choix,"est un excellent choix! Il coûte",magasin.get(choix))
    #calcul du montant total de l'achat
    n=int(input(print("entrer le nombre d'exemplaire en chiffre:")))
    MONTANT_ACHAT=n*magasin.get(choix)
    print("Le montant de votre achat s'élève à:",MONTANT_ACHAT)
    if MONTANT_ACHAT>5000:
        print("votre remise est de 10%")
        Total= MONTANT_ACHAT- MONTANT_ACHAT*0.1
        print("le montant à payer est:", Total)
    else:
        print("le montant à payer est:",MONTANT_ACHAT)
else:
    print("Nous ne disposons pas de ce sac en stock pour le moment")
```
